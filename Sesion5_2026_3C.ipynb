## 1. Conceptos de partida

La **estadística**, rama de las matemáticas, reúne procedimientos que permiten analizar correctamente los datos. Esos procedimientos —los **métodos estadísticos**— transforman los datos en **información útil** para tomar decisiones basadas en datos.

### 1.1 Dato, información y conocimiento

No son sinónimos: son tres peldaños de una misma escalera, y **el análisis de datos consiste en subirlos**.

| Peldaño | Qué es | Ejemplo |
|---|---|---|
| **Dato** | Registro crudo, sin contexto. | `27` |
| **Información** | El dato en su fila, respondiendo una pregunta. | *Yerson Arrieta anotó 27 goles: es el goleador.* |
| **Conocimiento** | Información que permite decidir. | *El cuerpo técnico arma el ataque alrededor de él.* |

### 1.2 Población y muestra

- **Población (N) — el universo:** conjunto total de individuos u objetos con características comunes que se desean estudiar. Es el **alcance máximo** de nuestras conclusiones.
- **Muestra (n) — el subconjunto:** segmento representativo extraído de la población. Es la **parte manejable y observable** del estudio.

**Ejemplo (Estudio 1).** Evaluar el conocimiento previo de los nuevos ingresos a la Facultad de Ingeniería:
- Población: los **2.500** estudiantes matriculados en primer semestre.
- Muestra: **120** estudiantes elegidos al azar que responden una prueba diagnóstica detallada.

### 1.3 El ciclo de la inferencia (y por qué no lo haremos aquí)

1. **Muestreo:** se selecciona una muestra a partir de la población.
2. **Análisis:** se obtienen estadísticos de la muestra.
3. **Inferencia:** los resultados se proyectan para concluir sobre la población.

> 📌 **En este curso no haremos inferencia.** Nos centraremos en **describir y resumir** datos —vengan de una población o de una muestra—: organizarlos, resumirlos y visualizarlos (tablas, medidas de tendencia central y de dispersión, gráficos), **sin generalizar** de una muestra a una población.

## 2. Datos tabulares, CSV y Pandas

### 2.1 ¿Qué son los datos tabulares?

Son datos organizados en una **tabla** de filas y columnas:

- **Columnas → variables o atributos** (por ejemplo, nombre, edad, nota).
- **Filas → registros u observaciones**: cada fila es un caso (un estudiante, una venta, un paciente… o, hoy, **un futbolista**).

Nuestro conjunto de datos describe a los jugadores del club ficticio **Promesas Bananeras**: **100 filas × 17 columnas**.

### 2.2 El formato CSV

Los datos de hoy vienen en un archivo **CSV** (*Comma-Separated Values*): texto plano con **valores separados por comas**. La primera línea trae los nombres de las columnas y cada línea siguiente es una fila.

### 2.3 Pandas

**Pandas** es el estándar de la industria para trabajar con datos tabulares en Python. Si un conjunto de datos puede manejarse en Excel o en un CSV, Pandas también puede manejarlo, y con más flexibilidad para el análisis.

| Hoy | **El estándar:** millones de personas la usan para analizar datos tabulares. |

## 5. Tipos de variables

Antes de transformar, hay que saber **qué clase de variable** es cada columna, porque eso decide qué operaciones tienen sentido.

```
Tipos de variables
├── Cuantitativas (números que se miden o cuentan)
│   ├── Discreta  → se cuenta, valores aislados   (goles, partidos jugados)
│   └── Continua  → se mide, admite decimales     (estatura, peso, salario)
└── Cualitativas (cualidades o categorías)
    ├── Nominal   → sin orden                      (nacionalidad, posición, pierna dominante)
    └── Ordinal   → con orden natural              (nivel de forma: Bajo < Medio < Alto < Élite)
```

Clasificación de las columnas de nuestro archivo:

| Tipo | Columnas |
|---|---|
| Cuantitativa discreta | `matches_played`, `goals`, `assists`, `yellow_cards`, `red_cards`, `joined_year`, `left_year` |
| Cuantitativa continua | `height_cm`, `weight_kg`, `salary_cop_million_month` |
| Cualitativa nominal | `nationality`, `position`, `dominant_foot` |
| Cualitativa ordinal | `fitness_level` |
| Identificadores y fecha | `player_id`, `full_name` (identifican, no se analizan) · `birth_date` (fecha) |

> 💡 Que una columna tenga números **no** la hace cuantitativa: un código postal o un número de camiseta son etiquetas. Promediarlos no tiene sentido.

## 6. Transformaciones de tipo

Ahora le decimos a Pandas el tipo correcto de cada columna. Cada transformación **reemplaza** la columna original (`df["col"] = ...`).

### 6.1 Texto → fecha

`pd.to_datetime()` convierte el texto a fecha. Con `errors="coerce"`, lo que no pueda leer se convierte en vacío (`NaT`) en lugar de detener el programa.

````python
df["birth_date"] = pd.to_datetime(df["birth_date"], errors="coerce")
df["birth_date"].head(3)
````

Con el tipo fecha ya podemos hacer cosas que con texto eran imposibles, como extraer el año de nacimiento:

````python
df["birth_date"].dt.year.head(3)
````

### 6.2 A enteros que admiten vacíos: `Int64`

Fíjate en la **I mayúscula**: `Int64` (de Pandas) es un entero que **sí admite vacíos**, a diferencia de `int64` (de NumPy). Por eso es el tipo adecuado para `left_year`: año entero, pero con jugadores sin año de retiro.

````python
df["left_year"] = pd.to_numeric(df["left_year"], errors="coerce").astype("Int64")
df["joined_year"] = df["joined_year"].astype("Int64")

df[["joined_year", "left_year"]].head(4)
````

### 6.3 A categóricos nominales

Para columnas de texto con **pocos valores que se repiten** y **sin orden**, usamos `category`. Ahorra memoria y deja explícito que la variable es cualitativa.

````python
df["nationality"] = df["nationality"].astype("category")
df["position"] = df["position"].astype("category")
df["dominant_foot"] = df["dominant_foot"].astype("category")

df["dominant_foot"].cat.categories
````

### 6.4 A categóricos ordinales

`fitness_level` tiene un **orden natural**: Bajo < Medio < Alto < Élite. Pandas no puede adivinarlo (alfabéticamente, «Alto» iría antes que «Bajo»), así que se lo damos explícitamente con `pd.Categorical(..., categories=..., ordered=True)`.

````python
fitness_order = ["Bajo", "Medio", "Alto", "Élite"]
df["fitness_level"] = pd.Categorical(df["fitness_level"],
                                     categories=fitness_order,
                                     ordered=True)

df["fitness_level"].head(5)
````

Observa la última línea de la salida: `['Bajo' < 'Medio' < 'Alto' < 'Élite']`. Ese orden permite **comparar**. Por ejemplo, ¿cuántos jugadores tienen nivel de forma Alto o superior?

````python
(df["fitness_level"] >= "Alto").sum()
````

### 6.5 Comprobar los cambios

Siempre que transformes, **verifica**. Compara con el `info()` anterior: `nationality`, `position`, `dominant_foot` y `fitness_level` ahora son `category`; `birth_date` es `datetime64[ns]`; `joined_year` y `left_year` son `Int64`.

````python
df.info()
````

## 7. Limpieza de datos

**Limpieza de datos:** operaciones que corrigen defectos de los datos crudos (faltantes, formatos inconsistentes, duplicados) para que el análisis sea **confiable**.

> Sin limpieza, la respuesta puede ser **correcta en el código y falsa en la realidad**.

Tres defectos típicos:

| # | Defecto | En nuestro archivo |
|---|---|---|
| 1 | **Faltantes:** celdas vacías que Pandas marca como `NaN`. | `left_year` tiene 58 vacíos. |
| 2 | **Texto inconsistente:** `"Brasil "` y `"brasil"` son dos países distintos para Pandas. | Se detecta con `unique()` y se corrige con `str.strip()`. |
| 3 | **Números como texto:** `"$6,36"` parece número, pero es una palabra. | No se puede sumar ni promediar hasta convertirlo. |

### 7.1 Valores faltantes: `NaN`

**NaN** (*Not a Number*) es la marca de Pandas para una celda vacía. **No es un cero: es ausencia de dato.**

- `isna()` marca cada celda: `True` si está vacía.
- `sum()` cuenta los `True` de cada columna (Python trata `True` como 1).

````python
df.isna().sum()
````

Solo `left_year` tiene vacíos: **58 de 100**.

> ⚠️ `errors="coerce"` convierte en vacío lo que no logra leer. **Después de transformar, vuelve a contar** los vacíos: si aumentaron, la transformación encontró valores que no entendió.

### 7.2 Un NaN también cuenta una historia

Antes de decidir qué hacer con un vacío, hay que preguntar **por qué está vacío**.

````python
df[["full_name", "joined_year", "left_year"]].head(4)
````

Con tipo `Int64`, el vacío se muestra como `<NA>`: es lo mismo que `NaN`.

**¿Por qué Jhon Suárez no tiene año de retiro?** Porque **sigue jugando** en el club. El vacío **es información**: 58 jugadores siguen activos.

> 🛠️ **Tip del ingeniero:** nunca borres una fila «porque estorba». Un vacío puede ser la huella de algo real: un jugador activo, un sensor apagado, alguien que prefirió no responder.

### 7.3 `dropna()`: apartar filas sin dato, con criterio

`dropna()` elimina las filas que tienen vacíos. Es rápido, y **por eso mismo es peligroso**.

- `subset` indica **en qué columnas** buscar vacíos. Sin `subset`, se borra la fila si tiene un vacío en **cualquier** columna.
- Devuelve un **DataFrame nuevo**: `df` conserva sus 100 filas.

````python
print("Antes:  ", df.shape)

df_retirados = df.dropna(subset=["left_year"])

print("Después:", df_retirados.shape)
print("df sigue intacto:", df.shape)
````

De 100 jugadores quedan **42**: los que **ya se fueron**. Desaparecen los **58 que siguen en el club**.

### 7.4 Ante un vacío hay tres opciones

Cada una tiene consecuencias. La decisión depende de **por qué está vacío**.

| Opción | Herramienta | Consecuencia |
|---|---|---|
| **Borrarlo** | `dropna()` | Se pierden filas y, con ellas, parte de la historia. En `left_year` se borran los 58 activos. |
| **Dejarlo** | `NaN` | Lo honesto cuando el vacío tiene significado, como en `left_year`: «aún juega». |
| **Rellenarlo** | `fillna()` | Solo si el valor es defendible. Inventar un dato es falsear la tabla. |

> 📏 **Regla práctica:** nunca borres ni rellenes sin poder explicar por qué.

En nuestro caso, **dejamos** los vacíos de `left_year` tal como están.

### 7.5 Texto inconsistente (1): verificar

Para una persona, `"Brasil"` y `"brasil"` son lo mismo. Para Pandas son **dos valores diferentes**: al contar, agrupar o filtrar, el país aparece partido.

````python
df["nationality"].nunique()   # número de valores distintos
````

````python
# Ordenamos la lista para que las variantes de un mismo país queden juntas
sorted(df["nationality"].unique())
````

### 7.6 Texto inconsistente (2): corregir

- `.str` aplica un método de texto a **toda** la columna.
- `strip()` quita espacios al inicio y al final.
- `title()` deja mayúscula inicial (`lower()` pasaría todo a minúsculas).

Como la columna ahora es texto limpio, la volvemos a convertir en `category`.

````python
df["nationality"] = (df["nationality"]
                     .str.strip()
                     .str.title()
                     .astype("category"))

df["nationality"].nunique()   # 16
````

````python
sorted(df["nationality"].unique())
````

## 8. Selecciones básicas

### 8.1 El ciclo de pregunta

Una **necesidad de datos** puede expresarse como una **pregunta** que se responde consultando lo almacenado. Por eso el código nunca aparece suelto: siempre lo pide una pregunta planteada antes.

| Paso | Qué se hace | Ejemplo |
|---|---|---|
| 1. Pregunta | En palabras de quien la necesita | «¿Nombre y posición de todos los jugadores?» |
| 2. Dato | ¿Qué columnas la responden? | `full_name` y `position` |
| 3. Operación | La herramienta de Pandas | Seleccionar columnas: `df[[…]]` |
| 4. Resultado | La tabla que se obtiene | 100 filas × 2 columnas |
| 5. Información | Lo que se concluye | El plantel completo, con su posición |

Avanzaremos así: **una herramienta nueva por ciclo**, en dificultad creciente. Al final de la sesión (sección 10) veremos el caso más exigente: cuando la necesidad llega **difusa** y primero hay que convertirla en pregunta.

### ❓ Pregunta 1: ¿Cuál es el nombre y la posición de todos los jugadores?

- **Dato:** `full_name` y `position`.
- **Operación:** seleccionar varias columnas → **corchetes dobles** con una lista de nombres.

````python
df[["full_name", "position"]]
````

**Información:** el plantel completo (100 filas × 2 columnas), con la posición de cada jugador.

### 8.2 ¿Corchetes simples o dobles?

La misma columna, pedida de dos formas, devuelve **dos estructuras distintas**.

````python
# Corchetes simples → Series (columna suelta)
print(type(df["full_name"]))
df["full_name"].head(3)
````

````python
# Corchetes dobles → DataFrame (tabla de una columna)
print(type(df[["full_name"]]))
df[["full_name"]].head(3)
````

### ❓ Pregunta 2: ¿Cuál es el nombre, la nacionalidad, la pierna dominante y la cantidad de goles de todos los jugadores?

*Los resultados se deben mostrar **ordenados de forma descendente** según la cantidad de goles.*

- **Operación nueva:** `sort_values(by=..., ascending=False)`.
- `by`: columna por la que se ordena. `ascending=False`: de mayor a menor.

````python
df[["full_name", "nationality", "dominant_foot", "goals"]].sort_values(
    by="goals",
    ascending=False
)
````

**Información:** Yerson Arrieta (Ecuador) encabeza con 27 goles, seguido por Jefferson Martínez (20) y Pablo Suárez (19).

### 8.3 Ordenar y seleccionar: el orden de los pasos

También se puede **ordenar primero y seleccionar después**. En este caso el resultado es idéntico:

````python
# A. Seleccionar → ordenar
resultado_A = (df[["full_name", "nationality", "dominant_foot", "goals"]]
               .sort_values(by="goals", ascending=False))

# B. Ordenar → seleccionar
resultado_B = (df.sort_values(by="goals", ascending=False)
               [["full_name", "nationality", "dominant_foot", "goals"]])

print("¿Son idénticos?", resultado_A.equals(resultado_B))
````

Pero **no siempre da igual**: depende de qué columnas quedan en la tabla al momento de ordenar.

**¿Cuándo sí importa?** Cuando se ordena por una columna que **no** se va a mostrar.
Pregunta: *nombre y goles, del salario más alto al más bajo.*

````python
# ✗ Seleccionar primero: el salario ya no está en la tabla cuando intentamos ordenar
try:
    df[["full_name", "goals"]].sort_values(by="salary_cop_million_month")
except KeyError as e:
    print("KeyError:", e)
````

````python
# ✓ Ordenar primero (con todas las columnas disponibles), recortar después
df.sort_values(by="salary_cop_million_month", ascending=False)[["full_name", "goals"]]
````

### 8.4 Al ordenar, cada fila conserva su índice

Mira la columna del índice en los resultados ordenados: 16, 52, 63, 47, 21… **No es un error.**

- Yerson Arrieta sigue siendo la **fila 16** aunque ahora aparezca de primera.
- El índice es la **identidad** de la fila, no su **posición**.
- Por defecto se ordena de menor a mayor (`ascending=True`).

````python
# Antes de ordenar: Yerson Arrieta está en la fila con etiqueta 16
df.loc[[0, 1, 2, 16], ["full_name", "goals"]]
````

````python
# Después de ordenar: aparece primero, pero conserva la etiqueta 16
df[["full_name", "goals"]].sort_values(by="goals", ascending=False).head(5)
````

*Mostrar los datos ordenados por la **cantidad de tarjetas**, del que más tarjetas ha tenido al que menos.*

No existe una columna «total de tarjetas»: hay que **crearla** sumando amarillas y rojas. Es una **columna calculada**.

````python
df["total_cards"] = df["yellow_cards"] + df["red_cards"]

df.sort_values(by="total_cards", ascending=False)[["full_name", "nationality"]]
````

Observa que ordenamos por `total_cards` **antes** de recortar columnas: si seleccionáramos primero, `total_cards` ya no estaría (como vimos con el salario).

### 8.6 Operación vectorizada

Al crear `total_cards` **no escribimos ningún bucle**: Pandas sumó las 100 filas de un solo golpe. Esto se llama **operación vectorizada**.

````python
df[["yellow_cards", "red_cards", "total_cards"]].head(3)
````

Lo que **evitamos** es algo así: un bucle que recorre fila por fila. Funciona, pero es más código, más lento y más fácil de equivocar.

````python
# Versión con bucle (solo para comparar; NO es la forma recomendada)
total = []
for i in range(len(df)):
    total.append(df["yellow_cards"][i] + df["red_cards"][i])

# ¿Da lo mismo que la versión vectorizada?
print("¿Mismo resultado?", list(df["total_cards"]) == total)
````

> **Columna calculada:** columna nueva cuyos valores se derivan de otras. No agrega datos del mundo: **revela información que ya estaba en la tabla**.

### 8.7 `assign()`: crear columnas sin tocar `df`

Dos formas de crear una columna calculada: **mismo resultado, distinto efecto** sobre `df`.

| Asignación directa | Con `assign()` |
|---|---|
| `df["total_cards"] = df["yellow_cards"] + df["red_cards"]` | `df_nuevo = df.assign(total_cards = df["yellow_cards"] + df["red_cards"])` |
| **Modifica** `df`: la columna queda agregada a la tabla original. | Devuelve un **DataFrame nuevo**: `df` queda exactamente igual. |

Para verlo con claridad, creamos con `assign()` una columna que `df` **no** tiene:

````python
df_nuevo = df.assign(goles_mas_asistencias = df["goals"] + df["assists"])

print("¿La tiene df_nuevo?", "goles_mas_asistencias" in df_nuevo.columns)
print("¿La tiene df?      ", "goles_mas_asistencias" in df.columns)
````

**Ventaja de `assign()`: se encadena.** Como devuelve una tabla nueva, se puede seguir operando sobre ella en la misma instrucción: **crear → ordenar → seleccionar → mostrar**, sin variables intermedias y sin modificar `df`.

> Los paréntesis externos permiten partir la instrucción en varias líneas, una por paso. Así se lee como una receta.

````python
(df.assign(total_cards = df["yellow_cards"] + df["red_cards"])
   .sort_values(by="total_cards", ascending=False)
   [["full_name", "total_cards"]]
   .head(3))
````

## 9. Filtrado de filas

### 9.1 Filtrar: quedarse solo con algunas filas

**Seleccionar** recorta **columnas**. **Filtrar** recorta **filas**: solo quedan las que responden «sí» a una condición.

| Seleccionar columnas | Filtrar filas |
|---|---|
| `df[["full_name", "goals"]]` | `df[df["goals"] > 10]` |

La condición es una **pregunta que se le hace a cada fila**: «¿tiene más de 10 goles?». Quedan las que responden `True`.

````python
df[df["goals"] > 10][["full_name", "goals"]]
````

### 9.2 Operadores de comparación

| Operador | Significa | Ejemplo |
|---|---|---|
| `==` | igual a | `df["nationality"] == "Nigeria"` |
| `!=` | distinto de | `df["position"] != "Portero"` |
| `>` | mayor que | `df["red_cards"] > 0` |
| `>=` | mayor o igual | `df["goals"] >= 5` |
| `<` | menor que | `df["weight_kg"] < 70` |
| `<=` | menor o igual | `df["goals"] <= 10` |

> ⚠️ Para **comparar** se usan dos signos igual (`==`); uno solo (`=`) **asigna**. Y el texto se compara **exacto**: `"Nigeria"` no es igual a `"nigeria"` ni a `"Nigeria "`.

### 9.3 La máscara booleana: una respuesta por fila

Una comparación **no filtra todavía**: le hace la pregunta a cada fila y devuelve una **Series de True/False**.

````python
condicion = df["red_cards"] > 0

# Ponemos lado a lado el dato y la respuesta para ver qué pasó
pd.DataFrame({"red_cards": df["red_cards"], "condicion": condicion}).head(8)
````

````python
print(type(condicion))
print("Cantidad de valores:", len(condicion))
````

`condicion` es una **Series**: solo índice y True/False, sin nombres ni tarjetas. Tiene **100 valores**, uno por jugador.

> **Definición. Máscara booleana:** Series de True/False que indica, para cada fila, si cumple una condición. **Filtrar** es entregarle esa máscara al DataFrame para quedarse con las filas `True`.

### 9.4 Contar y filtrar con la máscara

````python
# Contar cuántas cumplen: Python trata True como 1
condicion.sum()   # 41
````

````python
# Filtrar: la máscara va entre corchetes
df[condicion].shape   # 41 filas
````

**41 de los 100 jugadores** tienen al menos una tarjeta roja.

*Mostrar únicamente las primeras 5 filas.*

Hay **tres formas** de escribir el mismo filtro:

````python
# Alternativa 1: la máscara en una variable (la más fácil de leer al empezar)
condicion = df["red_cards"] > 0
df[condicion][["full_name", "nationality"]].head(5)
````

````python
# Alternativa 2: la condición dentro de los corchetes
df[df["red_cards"] > 0][["full_name", "nationality"]].head(5)
````

**¿Cuál usar?** Las tres son correctas. La 1 es la más fácil de leer al empezar. La 3 hace filas y columnas en un solo paso y es **la recomendada si luego se modifica el resultado**.

**Resultado:** observa el índice: salta de **7 a 10**. **El filtro no renumera**: cada fila conserva su etiqueta original, como al ordenar. Las filas 8 y 9 no aparecen porque esos jugadores tienen 0 tarjetas rojas.

### 9.5 `loc[filas, columnas]`: filtrar y seleccionar a la vez

`loc` recibe dos cosas separadas por una **coma**: qué filas y qué columnas.

```python
df.loc[df["red_cards"] > 0, ["full_name", "nationality"]]
#      └──── filas ──────┘  └────────── columnas ────────┘
```

| Antes de la coma: **filas** | Después de la coma: **columnas** |
|---|---|
| Una máscara: las filas donde es `True`. | Una lista de nombres: devuelve un **DataFrame**. |
| `:` significa todas las filas. | Un solo nombre, sin lista: devuelve una **Series** (o un valor). |
| Una etiqueta: `df.loc[7, ...]` es la fila 7. | `:` significa todas las columnas. |

**`loc` en acción: tres ejemplos.** En cada uno, lo que va antes de la coma decide las filas y lo que va después decide las columnas.

````python
# Máscara + todas las columnas → 41 filas (con todas las columnas)
df.loc[df["red_cards"] > 0, :].shape
````

````python
# Todas las filas + lista de columnas → 100 filas, 2 columnas
df.loc[:, ["full_name", "goals"]].shape
````

````python
# Una etiqueta de fila + un nombre de columna → un solo valor
df.loc[7, "full_name"]   # 'Diego Cárdenas'
````

### ❓ Pregunta 6: ¿Cuál es la nacionalidad, el nombre y la cantidad de goles de los jugadores que tienen entre 5 y 10 goles?

*Ordenar por nacionalidad ascendente y, dentro de cada nacionalidad, por goles descendente.*

Un rango son **dos condiciones que deben cumplirse a la vez** → se unen con `&` («y»). Cada condición va **entre paréntesis**.

````python
# Alternativa 1: dos condiciones con &
entre_5_y_10 = df.loc[
    (df["goals"] >= 5) & (df["goals"] <= 10),
    ["nationality", "full_name", "goals"]
].sort_values(
    by=["nationality", "goals"],
    ascending=[True, False])

entre_5_y_10
````

````python
# Alternativa 2: between(inicio, fin), incluye los extremos
alt_2 = df.loc[
    df["goals"].between(5, 10),
    ["nationality", "full_name", "goals"]
].sort_values(
    by=["nationality", "goals"],
    ascending=[True, False])

print("¿Mismo resultado?", entre_5_y_10.equals(alt_2))
print("Jugadores entre 5 y 10 goles:", len(alt_2))   # 34
````

- `between(5, 10)` **incluye los extremos**: equivale a `>= 5` y `<= 10`.
- Ordenar por dos columnas es lo que vimos antes: la segunda desempata dentro de la primera.

**Lectura del resultado:** Argentina no tiene jugadores con 8, 9 o 10 goles; por eso su lista empieza en 7. Brasil sí: arranca con los 8 goles de Gabriel Rodríguez.

### ❓ Pregunta 7: ¿Cuál es el nombre, la nacionalidad y el año de retiro de los jugadores de Venezuela o Ecuador que aún juegan en el club Promesas Bananeras?

*Lee la pregunta completa: tiene **más de una condición**.*

1. Nacionalidad **Ecuador o Venezuela** → `|` («o»).
2. **Aún juega** → un jugador activo no tiene año de retiro: su `left_year` está vacío → `df["left_year"].isna()`.
3. Las dos cosas a la vez → `&`.

````python
# Alternativa 1: | y & en la misma consulta
activos_ec_ve = df.loc[
    ((df["nationality"] == "Ecuador") | (df["nationality"] == "Venezuela"))
    & df["left_year"].isna(),
    ["full_name", "nationality", "left_year"]]

activos_ec_ve
````

````python
# Alternativa 2: isin() con una lista de valores posibles
alt_isin = df.loc[
    df["nationality"].isin(["Ecuador", "Venezuela"]) & df["left_year"].isna(),
    ["full_name", "nationality", "left_year"]]

print("¿Mismo resultado?", activos_ec_ve.equals(alt_isin))
print("Jugadores que aún juegan:", len(alt_isin))   # 12
````

**Cuidado con los paréntesis:** `&` se evalúa **antes** que `|`. Sin el paréntesis exterior, Pandas entiende «ecuatorianos, **o** venezolanos activos».

````python
# ✗ Sin el paréntesis exterior: la lógica cambia
mal = df.loc[
    (df["nationality"] == "Ecuador") | (df["nationality"] == "Venezuela")
    & df["left_year"].isna(),
    ["full_name", "nationality", "left_year"]]
print("Sin paréntesis exterior:", len(mal), "filas (se cuelan ecuatorianos retirados)")

# ✗ Olvidando la condición de 'aún juega'
sin_isna = df[df["nationality"].isin(["Ecuador", "Venezuela"])]
print("Sin isna():             ", len(sin_isna), "filas (se cuelan los que ya se fueron)")
````

Las dos alternativas correctas dan los mismos **12**. Sin el paréntesis exterior salen 18; sin `isna()`, 21. **El código corre en los tres casos: solo la lectura cuidadosa de la pregunta evita el error.**

### 9.6 `query()`: el filtro escrito casi en español

`query()` recibe la condición **como texto**. Dentro de ese texto sí se escriben `and`, `or` y `not`.

- Útil con varias condiciones, cuando los corchetes se vuelven difíciles de leer.
- Por dentro se usan **comillas simples**: `'Nigeria'`.
- **Solo filtra filas**; las columnas se eligen después.

````python
# Con máscara
df[(df["nationality"] == "Nigeria") & (df["salary_cop_million_month"] > 8)][
    ["full_name", "salary_cop_million_month"]]
````

````python
# Con query()
df.query("nationality == 'Nigeria' and salary_cop_million_month > 8")[
    ["full_name", "salary_cop_million_month"]]
````

## 10. Caso integrador: la pregunta manda, el código obedece

Hasta aquí las preguntas nos llegaron **ya bien formuladas**. En la práctica casi nunca es así: quien necesita el dato lo dice con sus palabras, y el primer trabajo es convertir esa necesidad en una pregunta que los datos puedan responder. Por eso este caso va al final: usa **todo** lo de la sesión —limpieza, columnas calculadas, ordenamiento y filtros— y además exige **criterio**.

### 10.1 De la necesidad difusa a la pregunta analizable

Primero se piensa **qué se quiere saber**; solo después se escribe código.

- **Necesidad difusa:** problema dicho en términos vagos; no dice qué medir ni con qué comparar. *«Siento que los fichajes recientes no están rindiendo.»*
- **Pregunta analizable:** dice qué variable se mide, sobre quiénes o en qué período, y contra qué se compara. *«¿Cuál es el promedio de goles por partido de los que llegaron en 2025, frente al de los que llegaron antes?»*

Delimitar es el primer trabajo del analista, **y el único que ningún programa hace por uno**.

### 10.2 Tres preguntas para delimitar una necesidad

| Criterio | Interrogatorio honesto | Decisión |
|---|---|---|
| 1. ¿Qué se mide? | ¿Qué es «rendir»? ¿Goles, asistencias, partidos? | Goles por partido: `goals ÷ matches_played` |
| 2. ¿Sobre quiénes? | ¿Quiénes cuentan como «recientes»? | Jugadores con `joined_year = 2025` |
| 3. ¿Comparado con qué? | «Menos» siempre es menos que algo. | Los que llegaron antes de 2025 |

Cada decisión se traduce en una herramienta que ya conocemos:

| Decisión | Herramienta |
|---|---|
| Goles por partido | Columna calculada (sección 8) |
| Llegaron en 2025 / antes de 2025 | Filtro con máscara booleana (sección 9) |
| Promedio de cada grupo | `.mean()` sobre una Series (sección 3) |

### 10.3 Paso 1: calcular lo que se mide

La decisión 1 pide **goles por partido** = `goals ÷ matches_played`. No existe esa columna: la creamos. El código es sencillo… pero mira el máximo y quién queda de primero al ordenar:

````python
df["goles_por_partido"] = df["goals"] / df["matches_played"]

df["goles_por_partido"].max()   # inf

df.sort_values(by="goles_por_partido", ascending=False)
````

### 10.4 Correcto en el código, falso en la realidad

¡`inf` (infinito)! El código está bien; **el problema está en el dato**.

**1. Detectarlo:** ¿cuántos jugadores tienen 0 partidos jugados?

````python
(df["matches_played"] == 0).sum()   # 1
````

````python
df.loc[df["matches_played"] == 0, ["full_name", "goals", "matches_played"]]
````

Brayan Figueroa tiene **3 goles y 0 partidos jugados**: dividir entre cero produce infinito.

**2. Entenderlo:** ¿error de captura? ¿registro incompleto? **Se pregunta a quien entregó los datos; no se adivina.**

**3. Decidir y avisar:** mientras se aclara, esa fila **no entra al ranking**, y al presentar resultados se informa cuál se apartó y por qué.

````python
ranking = (df[df["matches_played"] > 0]
           .sort_values(by="goles_por_partido", ascending=False)
           [["full_name", "goals", "matches_played", "goles_por_partido"]])
ranking.head(5)
````

### 10.5 Paso 2: responder la pregunta analizable

Ahora sí aplicamos las decisiones 2 y 3 con los **filtros** de la sección 9, trabajando solo con los registros válidos (sin el jugador de 0 partidos):

````python
validos = df[df["matches_played"] > 0]

recientes = validos[validos["joined_year"] == 2025]["goles_por_partido"].mean()
anteriores = validos[validos["joined_year"] < 2025]["goles_por_partido"].mean()

print(f"Promedio de goles por partido, llegados en 2025:     {recientes:.3f}")
print(f"Promedio de goles por partido, llegados antes de 2025: {anteriores:.3f}")
print("Nota: se excluyó 1 jugador con 0 partidos jugados (Brayan Figueroa) hasta aclarar su registro.")
````

### 10.6 De la respuesta al conocimiento

Recorramos el **ciclo de pregunta** con este caso:

| Paso | En este caso |
|---|---|
| 1. Pregunta | «¿Cuál es el promedio de goles por partido de los que llegaron en 2025, frente al de los que llegaron antes?» |
| 2. Dato | `goals`, `matches_played`, `joined_year` |
| 3. Operación | Columna calculada → apartar el registro con 0 partidos → dos filtros → dos promedios |
| 4. Resultado | Dos números: el promedio de los llegados en 2025 y el de los llegados antes |
| 5. Información | Si el primero es menor, los datos **respaldan** la sensación inicial; si no, la **contradicen** |
