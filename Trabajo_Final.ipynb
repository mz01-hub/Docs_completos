### 2.2 Crear la copia de trabajo y retirar duplicados

Primero se hace una copia para no modificar \`df_encuesta\`. Después se eliminan únicamente las filas completamente idénticas. Así se conserva la primera aparición de cada respuesta.

```python
filas_originales = len(df_encuesta)
duplicados_exactos = df_encuesta.duplicated().sum()

df_normalizada = df_encuesta.drop_duplicates(keep="first").copy()

print("Filas originales:", filas_originales)
print("Copias duplicadas retiradas:", duplicados_exactos)
print("Dimensiones después de retirar duplicados:", df_normalizada.shape)
print("User_ID repetidos que quedan:", df_normalizada["User_ID"].duplicated().sum())
```

### 2.3 Estandarizar las categorías

Se quitan espacios sobrantes y se reemplazan las variantes que ya se detectaron en el Producto 1. Los nombres finales se escriben de forma uniforme para que las combinaciones con los datos de herramientas y países encuentren coincidencias.

No se rellenan las celdas vacías. Solo se cambian los textos para los que existe una equivalencia clara.

```python
columnas_texto = ["Gender", "Country", "Education_Level", "AI_Tool"]

for columna in columnas_texto:
    df_normalizada[columna] = df_normalizada[columna].str.strip()

mapa_genero = {
    "M": "Male",
    "male": "Male",
    "MALE": "Male",
    "F": "Female",
    "female": "Female",
    "FEMALE": "Female",
    "other": "Other"
}

mapa_pais = {
    "australia": "Australia",
    "brazil": "Brazil",
    "canada": "Canada",
    "france": "France",
    "india": "India",
    "philippines": "Philippines",
    "U.K.": "United Kingdom",
    "United States of America": "United States"
}

mapa_educacion = {
    "BA/BSc": "Bachelor's",
    "bachelor's degree": "Bachelor's",
    "Bachelors": "Bachelor's",
    "high school": "High School",
    "HS": "High School",
    "MA/MSc": "Master's",
    "master's degree": "Master's",
    "Masters": "Master's",
    "Ph.D.": "PhD",
    "undergrad": "Undergraduate"
}

mapa_herramientas = {
    "Chat GPT": "ChatGPT",
    "chatgpt": "ChatGPT",
    "Chatgpt": "ChatGPT",
    "claude": "Claude",
    "gemini": "Google Gemini",
    "Gemini": "Google Gemini",
    "Copilot": "Microsoft Copilot",
    "copilot": "Microsoft Copilot"
}

df_normalizada["Gender"] = df_normalizada["Gender"].replace(mapa_genero)
df_normalizada["Country"] = df_normalizada["Country"].replace(mapa_pais)
df_normalizada["Education_Level"] = df_normalizada["Education_Level"].replace(mapa_educacion)
df_normalizada["AI_Tool"] = df_normalizada["AI_Tool"].replace(mapa_herramientas)

print("Valores de Gender:", sorted(df_normalizada["Gender"].dropna().unique()))
print("Valores de Country:", sorted(df_normalizada["Country"].dropna().unique()))
print("Valores de Education_Level:", sorted(df_normalizada["Education_Level"].dropna().unique()))
print("Valores de AI_Tool:", sorted(df_normalizada["AI_Tool"].dropna().unique()))
```

### 2.4 Convertir números y tratar los valores fuera de rango

Algunos números llegaron como texto porque tenían unidades o símbolos. La función \`a_numero()\` retira esos caracteres y usa \`pd.to_numeric()\` para convertir la columna. Después se aplican los rangos establecidos en el diagnóstico.

Cuando un número queda fuera del rango, se cambia solamente esa celda a \`NaN\`; no se elimina toda la fila.

```python
columnas_numericas_texto = [
    "Age", "Monthly_Income", "AI_Usage_Hours_Per_Day",
    "Monthly_AI_Cost", "Time_Saved_Hours_Per_Week",
    "Work_or_Study_Hours_Per_Day"
]

def a_numero(serie):
    limpio = (serie.str.replace("yrs", "", regex=False)
                    .str.replace("hrs", "", regex=False)
                    .str.replace("$", "", regex=False)
                    .str.replace(",", "", regex=False)
                    .str.strip())
    return pd.to_numeric(limpio, errors="coerce")

for columna in columnas_numericas_texto:
    df_normalizada[columna] = a_numero(df_normalizada[columna])

rangos_validos = {
    "Age": (16, 80),
    "Monthly_Income": (0, 20000),
    "AI_Usage_Hours_Per_Day": (0, 16),
    "Work_or_Study_Hours_Per_Day": (0, 16),
    "Time_Saved_Hours_Per_Week": (0, 168),
    "Productivity_Score": (1, 10),
    "Accuracy_Rating": (1, 5),
    "Satisfaction_Score": (1, 10)
}

for columna, limites in rangos_validos.items():
    fuera_de_rango = (df_normalizada[columna].notna() &
                      ~df_normalizada[columna].between(limites[0], limites[1]))
    print(columna, "- valores fuera de rango:", fuera_de_rango.sum())
    df_normalizada.loc[fuera_de_rango, columna] = np.nan

costos_negativos = (df_normalizada["Monthly_AI_Cost"].notna() &
                    (df_normalizada["Monthly_AI_Cost"] < 0))
print("Monthly_AI_Cost - valores negativos:", costos_negativos.sum())
df_normalizada.loc[costos_negativos, "Monthly_AI_Cost"] = np.nan
```

### 2.5 Crear la columna \`Num_Tareas\`

En \`Tasks_Performed\` las tareas están separadas por comas. Se divide el texto por ese separador y se cuenta cuántos fragmentos hay.

Cuando \`Tasks_Performed\` está vacío, \`Num_Tareas\` también queda como faltante. No se asigna cero porque no sabemos cuántas tareas realiza esa persona.

```python
df_normalizada["Num_Tareas"] = (
    df_normalizada["Tasks_Performed"]
    .str.split(",")
    .str.len()
    .astype("Int64")
)

print("Cantidad de tareas registradas:")
print(df_normalizada["Num_Tareas"].value_counts(dropna=False).sort_index())
```

### 2.6 Crear la columna \`Area_Profesional\`

La profesión es texto libre y aparece en varios idiomas. Para resumirla, se crean familias a partir de palabras clave con \`str.contains()\`.

**Orden de las reglas:** primero se identifican los estudiantes; luego las profesiones de desarrollo, datos, atención al cliente, investigación, contenido y administración. El orden evita que un estudiante sea clasificado como profesional por una palabra que aparezca en su descripción.

Las profesiones que no coincidan con ninguna regla quedan en **Otra**. Cuando \`Profession\` está vacío, \`Area_Profesional\` también queda como faltante: no se inventa una profesión.

```python
profesion = df_normalizada["Profession"].str.strip()

es_estudiante = profesion.str.contains(
    r"\bstudent\b", case=False, na=False
)

es_estudiante_ingenieria = profesion.str.contains(
    r"computer science|software|engineering|engineer|information technology|\bIT\b|machine learning|artificial intelligence|data science|electronics|computing|programming",
    case=False, na=False
)

df_normalizada["Area_Profesional"] = "Otra"

# Los estudiantes se clasifican antes que las profesiones.
df_normalizada.loc[
    es_estudiante & es_estudiante_ingenieria,
    "Area_Profesional"
] = "Estudiante de ingeniería y computación"

df_normalizada.loc[
    es_estudiante & ~es_estudiante_ingenieria,
    "Area_Profesional"
] = "Estudiante de otras áreas"

no_estudiante = profesion.notna() & ~es_estudiante

# 1. Desarrollo de software
mascara = no_estudiante & profesion.str.contains(
    r"software\s+(developer|engineer|entwickler|entwicklerin)|softwareentwickler(in)?|developer|desenvolvedor|desenvolvedora|développeur|logiciel|full\s*stack|tech lead|technical lead|machine learning engineer|systems developer",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Desarrollo de software"

# 2. Análisis de datos
mascara = no_estudiante & profesion.str.contains(
    r"data analyst|data scientist|data science|data analytics|business intelligence|business analytics|analyste de données|analista de dados|datenanalyst|analytics",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Análisis de datos"

# 3. Atención al cliente
mascara = no_estudiante & profesion.str.contains(
    r"customer|support|service|success|kundenservice|kundenbetreuer|atendimento",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Atención al cliente"

# 4. Investigación y academia
mascara = no_estudiante & profesion.str.contains(
    r"research|professor|academic|scientist|researcher|pesquisador|phd",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Investigación y academia"

# 5. Contenido y marketing
mascara = no_estudiante & profesion.str.contains(
    r"content|contenu|writer|writing|marketing|social media|digital media|multimedia|film production|gestor de conteúdo|copywriting",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Contenido y marketing"

# 6. Administración y negocios
mascara = no_estudiante & profesion.str.contains(
    r"administrat|business analyst|business administration|project manager|projektmanager|data entry|assistente administrativo",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Administración y negocios"

# Los cargos sin coincidencia quedan como "Otra".
# Los 14 valores vacíos de Profession se conservan como faltantes.
df_normalizada.loc[
    df_normalizada["Profession"].isna(),
    "Area_Profesional"
] = np.nan

categorias_area = [
    "Desarrollo de software",
    "Análisis de datos",
    "Contenido y marketing",
    "Atención al cliente",
    "Administración y negocios",
    "Investigación y academia",
    "Estudiante de ingeniería y computación",
    "Estudiante de otras áreas",
    "Otra",
    "(sin información)"
]

frecuencia_areas = (
    df_normalizada["Area_Profesional"]
    .value_counts(dropna=False)
    .reindex(categorias_area, fill_value=0)
    .to_frame("Registros")
)
frecuencia_areas["% del total"] = (
    frecuencia_areas["Registros"] / len(df_normalizada) * 100
).round(1)

display(frecuencia_areas)

registros_otra = df_normalizada["Area_Profesional"].eq("Otra").sum()
porcentaje_otra = registros_otra / len(df_normalizada) * 100

print("Registros en Otra:", registros_otra)
print(f"Porcentaje de registros en Otra: {porcentaje_otra:.1f}%")
print("Profesiones sin información:", df_normalizada["Profession"].isna().sum())
```

### 3.2 Declarar las categorías nominales y ordinales

Las categorías nominales no tienen un orden. Por eso se guardan como \`category\` con la lista completa de opciones. Si una categoría no tuvo respuestas, puede seguir apareciendo con frecuencia cero.

Las categorías ordinales sí tienen un orden. Se usa \`ordered=True\` para que pandas respete ese orden al comparar o resumir los niveles.

```python
df_tipada = df_normalizada.copy()

# Texto e identificadores.
for columna in ["User_ID", "Profession", "Tasks_Performed"]:
    df_tipada[columna] = df_tipada[columna].astype("string")

# Cantidades enteras que pueden tener valores faltantes.
for columna in ["Age", "Num_Tareas", "Satisfaction_Score"]:
    df_tipada[columna] = pd.to_numeric(
        df_tipada[columna], errors="coerce"
    ).astype("Int64")

# Cantidades que pueden tener decimales.
columnas_continuas = [
    "Monthly_Income",
    "AI_Usage_Hours_Per_Day",
    "Monthly_AI_Cost",
    "Productivity_Score",
    "Time_Saved_Hours_Per_Week",
    "Work_or_Study_Hours_Per_Day"
]

for columna in columnas_continuas:
    df_tipada[columna] = pd.to_numeric(
        df_tipada[columna], errors="coerce"
    ).astype("float64")

# La calificación de exactitud se declara como una escala ordenada.
df_tipada["Accuracy_Rating"] = pd.to_numeric(
    df_tipada["Accuracy_Rating"], errors="coerce"
).astype("Int64")

# Listas completas de categorías nominales.
categorias_genero = ["Female", "Male", "Non-binary", "Other"]
categorias_pais = df_paises["pais"].dropna().drop_duplicates().tolist()
categorias_usuario = ["Professional", "Student"]
categorias_herramienta = (
    df_herramientas["Herramienta"].dropna().drop_duplicates().tolist()
)
categorias_proposito = [
    "Coding",
    "Content Creation",
    "Customer Support",
    "Data Analysis",
    "Office Work",
    "Research",
    "Studying",
    "Writing"
]
categorias_area = [
    "Desarrollo de software",
    "Análisis de datos",
    "Contenido y marketing",
    "Atención al cliente",
    "Administración y negocios",
    "Investigación y academia",
    "Estudiante de ingeniería y computación",
    "Estudiante de otras áreas",
    "Otra"
]

df_tipada["Gender"] = pd.Categorical(
    df_tipada["Gender"], categories=categorias_genero
)
df_tipada["Country"] = pd.Categorical(
    df_tipada["Country"], categories=categorias_pais
)
df_tipada["User_Type"] = pd.Categorical(
    df_tipada["User_Type"], categories=categorias_usuario
)
df_tipada["AI_Tool"] = pd.Categorical(
    df_tipada["AI_Tool"], categories=categorias_herramienta
)
df_tipada["AI_Purpose"] = pd.Categorical(
    df_tipada["AI_Purpose"], categories=categorias_proposito
)
df_tipada["Area_Profesional"] = pd.Categorical(
    df_tipada["Area_Profesional"], categories=categorias_area
)

# Variables ordinales: el orden se escribe de menor a mayor.
orden_educacion = [
    "High School",
    "Diploma",
    "Undergraduate",
    "Bachelor's",
    "Master's",
    "PhD"
]
df_tipada["Education_Level"] = pd.Categorical(
    df_tipada["Education_Level"],
    categories=orden_educacion,
    ordered=True
)

df_tipada["Accuracy_Rating"] = pd.Categorical(
    df_tipada["Accuracy_Rating"],
    categories=[1, 2, 3, 4, 5],
    ordered=True
)

df_tipada["Would_Recommend"] = pd.Categorical(
    df_tipada["Would_Recommend"],
    categories=["No", "Maybe", "Yes"],
    ordered=True
)

print("Categorías de herramientas:", df_tipada["AI_Tool"].cat.categories.tolist())
print("Categorías de países:", df_tipada["Country"].cat.categories.tolist())
print("Orden de Education_Level:", df_tipada["Education_Level"].cat.categories.tolist())
print("Orden de Accuracy_Rating:", df_tipada["Accuracy_Rating"].cat.categories.tolist())
print("Orden de Would_Recommend:", df_tipada["Would_Recommend"].cat.categories.tolist())
```

### 3.3 ¿Por qué se eligió cada tipo?

- User_ID se guarda como texto porque identifica una respuesta. No tiene sentido sumarlo ni calcular un promedio.
- Age y Num_Tareas son enteros porque cuentan años cumplidos y tareas. Se usa \`Int64\` porque permite conservar los faltantes.
- Las horas, el ingreso y los costos se guardan como números decimales porque pueden tener partes decimales y se pueden resumir con operaciones matemáticas.
- Gender, Country, User_Type, AI_Tool, AI_Purpose y Area_Profesional son categorías sin orden. Una categoría no es mayor que otra.
- Education_Level, Accuracy_Rating y Would_Recommend tienen un orden. Por eso se declaran con \`ordered=True\`.
- Profession y Tasks_Performed son texto libre. No se convierten en categorías porque hay muchas formas distintas de escribirlas.
- Satisfaction_Score se guarda como entero porque registra una puntuación numérica entre 1 y 10. En este análisis se cuenta como una cantidad discreta.
- Productivity_Score se conserva como decimal porque en la encuesta aparecen valores como 5.3 y 9.9.

Pandas indica cómo está guardado el dato. La clasificación estadística depende de lo que representa en la encuesta.

#### Necesidad 2. Satisfacción según la herramienta

Pregunta: ¿Qué herramientas tienen mayor satisfacción promedio y cuántas respuestas respaldan cada promedio?

Herramientas: selección de columnas, groupby, agg y sort_values.

## Producto 5. Tablas de frecuencia

Se utilizan las 288 respuestas sin duplicados. En cada tabla, el total (n) corresponde a las respuestas válidas de la variable o del cruce; los datos faltantes no se cuentan como una categoría. La función `tabla_frecuencias()` reconoce si la variable es nominal, ordinal o numérica: las nominales no llevan acumuladas y las ordinales y cuantitativas sí.

Las tablas por intervalos usan la convención `[a, b)`: incluyen el límite izquierdo y excluyen el derecho. La última clase de ingresos queda abierta por la derecha.

### 5.1 Función para crear tablas de frecuencia

La función conserva todas las categorías declaradas, incluso cuando alguna tiene frecuencia cero. En variables numéricas permite trabajar con valores individuales o con intervalos. Para los intervalos calcula la marca de clase cuando los dos límites son finitos.

```python
def tabla_frecuencias(
    serie,
    tipo="auto",
    bins=None
):
    """
    Crea una tabla de frecuencias.
    tipo: auto, nominal, ordinal, discreta o intervalos.
    Las nominales no llevan Fi ni Hi.
    """
    es_categorica = isinstance(serie.dtype, pd.CategoricalDtype)

    if tipo == "auto":
        if es_categorica and serie.cat.ordered:
            tipo = "ordinal"
        elif es_categorica:
            tipo = "nominal"
        elif pd.api.types.is_numeric_dtype(serie):
            tipo = "discreta"
        else:
            tipo = "nominal"

    datos = serie.dropna()
    n = len(datos)

    if tipo in ["nominal", "ordinal", "discreta"]:
        if es_categorica:
            categorias = list(serie.cat.categories)
        else:
            categorias = sorted(datos.unique().tolist())

        frecuencias = (
            datos.value_counts(sort=False)
            .reindex(categorias, fill_value=0)
        )

        tabla = pd.DataFrame({
            "Valor": categorias,
            "fi": frecuencias.to_numpy()
        })

        tabla["hi"] = tabla["fi"] / n if n else 0
        tabla["%"] = tabla["hi"] * 100

        if tipo != "nominal":
            tabla["Fi"] = tabla["fi"].cumsum()
            tabla["Hi"] = tabla["hi"].cumsum()

        tabla.attrs["n"] = n
        tabla.attrs["tipo"] = tipo
        return tabla

    if tipo == "intervalos":
        if bins is None:
            raise ValueError("Para intervalos se debe indicar bins.")

        agrupados = pd.cut(
            datos,
            bins=bins,
            right=False,
            include_lowest=True
        )
        intervalos = agrupados.cat.categories
        frecuencias = agrupados.value_counts(sort=False)

        nombres = []
        marcas = []
        for intervalo in intervalos:
            inferior = float(intervalo.left)
            superior = float(intervalo.right)
            if np.isfinite(superior):
                nombres.append(f"[{inferior:g}, {superior:g})")
                marcas.append((inferior + superior) / 2)
            else:
                nombres.append(f"[{inferior:g}, infinito)")
                marcas.append(np.nan)

        tabla = pd.DataFrame({
            "Intervalo": nombres,
            "Marca_de_clase": marcas,
            "fi": frecuencias.to_numpy()
        })
        tabla["hi"] = tabla["fi"] / n if n else 0
        tabla["%"] = tabla["hi"] * 100
        tabla["Fi"] = tabla["fi"].cumsum()
        tabla["Hi"] = tabla["hi"].cumsum()
        tabla.attrs["n"] = n
        tabla.attrs["tipo"] = tipo
        return tabla

    raise ValueError("Tipo de variable no reconocido.")


def verificar_suma_frecuencias(nombre, tabla, serie):
    n_valido = int(serie.notna().sum())
    suma_fi = int(tabla["fi"].sum())
    print(f"{nombre}: Σfi = {suma_fi}; n válido = {n_valido}")
    assert suma_fi == n_valido, "La suma de fi no coincide con n."
    print("Verificación correcta.")
```

### 5.2 Tabla simple de una variable nominal: género

El género es nominal porque sus categorías no tienen un orden. Por eso la tabla presenta (fi), (hi) y porcentaje, pero no (Fi) ni (Hi). Se conservan las cuatro categorías definidas en la encuesta, aunque alguna no tenga respuestas.

**Interpretación:** permite identificar qué categoría reúne más respuestas y qué proporción representa cada una. La tabla no establece una jerarquía entre categorías.

```python
tabla_genero = tabla_frecuencias(df_analisis["Gender"])

display(tabla_genero)
verificar_suma_frecuencias(
    "Género", tabla_genero, df_analisis["Gender"]
)

fila_mas_frecuente = tabla_genero.loc[tabla_genero["fi"].idxmax()]
porcentaje_male = tabla_genero.loc[
    tabla_genero["Valor"] == "Male", "%"
].iloc[0]

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Cuál categoría es la más frecuente? "
    f"{fila_mas_frecuente['Valor']} "
    f"({int(fila_mas_frecuente['fi'])} respuestas)."
)
print(
    "2. ¿Qué porcentaje corresponde a la categoría Male? "
    f"{porcentaje_male:.1f}%."
)
```

### 5.3 Tabla simple de una variable ordinal: disposición a recomendar

`Would_Recommend` tiene el orden No, Maybe y Yes. Se conservan ese orden y las categorías sin respuestas. Como existe un orden, se calculan también las frecuencias acumuladas (Fi) y (Hi).

**Interpretación:** la frecuencia acumulada permite saber qué parte respondió hasta una categoría determinada. Por ejemplo, se puede reunir No y Maybe para conocer la proporción que no respondió Yes.

```python
tabla_recomendacion = tabla_frecuencias(
    df_analisis["Would_Recommend"]
)

display(tabla_recomendacion)
verificar_suma_frecuencias(
    "Disposición a recomendar",
    tabla_recomendacion,
    df_analisis["Would_Recommend"]
)

n_recomendacion = tabla_recomendacion["fi"].sum()
n_yes = tabla_recomendacion.loc[
    tabla_recomendacion["Valor"] == "Yes", "fi"
].iloc[0]
n_no_maybe = tabla_recomendacion.loc[
    tabla_recomendacion["Valor"].isin(["No", "Maybe"]), "fi"
].sum()

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Qué porcentaje respondió Yes? "
    f"{n_yes / n_recomendacion * 100:.1f}%."
)
print(
    "2. ¿Qué porcentaje respondió No o Maybe? "
    f"{n_no_maybe / n_recomendacion * 100:.1f}%."
)
```

### 5.4 Primera tabla simple cuantitativa discreta: número de tareas

`Num_Tareas` es discreta porque cuenta tareas y toma valores enteros. La tabla ordena los valores de menor a mayor e incluye las frecuencias acumuladas.

**Interpretación:** permite observar cuántas tareas distintas reporta cada persona y dónde se concentra la mayor frecuencia. La acumulada ayuda a responder preguntas como qué porcentaje realiza tres tareas o menos.

```python
tabla_tareas = tabla_frecuencias(
    df_analisis["Num_Tareas"],
    tipo="discreta"
)

display(tabla_tareas)
verificar_suma_frecuencias(
    "Número de tareas",
    tabla_tareas,
    df_analisis["Num_Tareas"]
)

n_tareas = tabla_tareas["fi"].sum()
porcentaje_tres_mas = (
    tabla_tareas.loc[tabla_tareas["Valor"] >= 3, "fi"].sum()
    / n_tareas * 100
)
fila_moda_tareas = tabla_tareas.loc[tabla_tareas["fi"].idxmax()]

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Qué porcentaje registra tres tareas o más? "
    f"{porcentaje_tres_mas:.1f}%."
)
print(
    "2. ¿Cuántas tareas aparecen con mayor frecuencia? "
    f"{fila_moda_tareas['Valor']} tareas "
    f"({int(fila_moda_tareas['fi'])} respuestas)."
)
```

```python
tabla_satisfaccion = tabla_frecuencias(
    df_analisis["Satisfaction_Score"],
    tipo="discreta"
)

display(tabla_satisfaccion)
verificar_suma_frecuencias(
    "Satisfacción",
    tabla_satisfaccion,
    df_analisis["Satisfaction_Score"]
)

n_satisfaccion = tabla_satisfaccion["fi"].sum()
n_satisfaccion_baja = tabla_satisfaccion.loc[
    tabla_satisfaccion["Valor"] <= 6, "fi"
].sum()
fila_moda_satisfaccion = tabla_satisfaccion.loc[
    tabla_satisfaccion["fi"].idxmax()
]

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Qué porcentaje calificó la satisfacción con 6 o menos? "
    f"{n_satisfaccion_baja / n_satisfaccion * 100:.1f}%."
)
print(
    "2. ¿Cuál puntuación tiene la mayor frecuencia? "
    f"{fila_moda_satisfaccion['Valor']} "
    f"({int(fila_moda_satisfaccion['fi'])} respuestas)."
)
```

### 5.6 Tabla por intervalos con la regla de Sturges: edad

Se usa la regla de Sturges (k = 1 + 3.322\log_{10}(n)) para sugerir el número de clases. Se redondea al entero más cercano. Los límites se distribuyen desde la edad mínima hasta un valor apenas superior a la máxima para que el último dato también quede incluido en la convención `[a, b)`.

**Interpretación:** agrupar edades facilita ver en qué tramo se concentran las personas. La marca de clase representa el punto medio de cada intervalo y sirve para aproximar la media agrupada en el Producto 7.

```python
edades_validas = df_analisis["Age"].dropna()
n_edades = len(edades_validas)

k_sturges = int(round(1 + 3.322 * np.log10(n_edades)))
edad_min = edades_validas.min()
edad_max = edades_validas.max()
limite_final = np.nextafter(float(edad_max), np.inf)
bins_edad_sturges = np.linspace(
    float(edad_min), limite_final, k_sturges + 1
)

tabla_edad_sturges = tabla_frecuencias(
    df_analisis["Age"],
    tipo="intervalos",
    bins=bins_edad_sturges
)

print("Número de clases sugerido por Sturges:", k_sturges)
display(tabla_edad_sturges)
verificar_suma_frecuencias(
    "Edad agrupada por Sturges",
    tabla_edad_sturges,
    df_analisis["Age"]
)

clase_mediana_edad = tabla_edad_sturges.loc[
    tabla_edad_sturges["Fi"] >= n_edades / 2
].iloc[0]
clase_moda_edad = tabla_edad_sturges.loc[
    tabla_edad_sturges["fi"].idxmax()
]

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿En qué intervalo se alcanza el 50% acumulado? "
    f"{clase_mediana_edad['Intervalo']} "
    f"(Hi = {clase_mediana_edad['Hi'] * 100:.1f}%)."
)
print(
    "2. ¿Cuál es el intervalo con más personas? "
    f"{clase_moda_edad['Intervalo']} "
    f"({int(clase_moda_edad['fi'])} personas)."
)
```

### 5.7 Tabla por intervalos con límite inicial y amplitud del equipo: horas de uso diario

Para las horas de uso diario se fija el límite inicial en 0 y la amplitud en 2 horas. Se forman clases `[0, 2)`, `[2, 4)` y así sucesivamente. Esta elección facilita relacionar la tabla con los grupos de intensidad que se usaron en el Producto 4.

**Interpretación:** la tabla muestra cuántas personas están en cada tramo de uso diario. La frecuencia acumulada permite conocer el porcentaje que usa la IA menos de cierta cantidad de horas.

```python
limite_inicial_uso = 0
amplitud_uso = 2
maximo_uso = df_analisis["AI_Usage_Hours_Per_Day"].max()
limite_superior_uso = (
    np.floor(maximo_uso / amplitud_uso) + 1
) * amplitud_uso

bins_uso = np.arange(
    limite_inicial_uso,
    limite_superior_uso + amplitud_uso,
    amplitud_uso
)

tabla_uso_intervalos = tabla_frecuencias(
    df_analisis["AI_Usage_Hours_Per_Day"],
    tipo="intervalos",
    bins=bins_uso
)

print("Límite inicial:", limite_inicial_uso)
print("Amplitud de clase:", amplitud_uso, "horas")
display(tabla_uso_intervalos)
verificar_suma_frecuencias(
    "Horas de uso diario",
    tabla_uso_intervalos,
    df_analisis["AI_Usage_Hours_Per_Day"]
)

uso_valido = df_analisis["AI_Usage_Hours_Per_Day"].dropna()
n_uso = len(uso_valido)
porcentaje_uso_4_mas = uso_valido.ge(4).mean() * 100
clase_moda_uso = tabla_uso_intervalos.loc[
    tabla_uso_intervalos["fi"].idxmax()
]

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Qué porcentaje usa IA cuatro horas o más al día? "
    f"{porcentaje_uso_4_mas:.1f}%."
)
print(
    "2. ¿Qué intervalo concentra más respuestas? "
    f"{clase_moda_uso['Intervalo']} "
    f"({int(clase_moda_uso['fi'])} respuestas)."
)
```

### 5.8 Tabla por intervalos con último intervalo abierto: ingreso mensual

Para el ingreso mensual se usan intervalos de amplitud 2 000 USD, desde 0. El último intervalo se deja abierto a la derecha: `[10 000, infinito)`. Esta clase evita fijar un límite máximo arbitrario y permite registrar cualquier ingreso válido que supere los 10 000 USD.

En una clase abierta no existe un punto medio matemático, por lo que su marca de clase aparece como no disponible. Las demás clases sí muestran su marca.

**Interpretación:** la tabla permite ubicar los ingresos por tramos y reconocer en qué tramo se agrupan más respuestas. Como hay valores faltantes, el denominador es el número de ingresos válidos, no las 288 filas.

```python
bins_ingreso = [0, 2000, 4000, 6000, 8000, 10000, np.inf]

tabla_ingreso_intervalos = tabla_frecuencias(
    df_analisis["Monthly_Income"],
    tipo="intervalos",
    bins=bins_ingreso
)

display(tabla_ingreso_intervalos)
verificar_suma_frecuencias(
    "Ingreso mensual",
    tabla_ingreso_intervalos,
    df_analisis["Monthly_Income"]
)

ingreso_valido = df_analisis["Monthly_Income"].dropna()
n_ingreso = len(ingreso_valido)
porcentaje_ingreso_8000_mas = ingreso_valido.ge(8000).mean() * 100
ultima_clase_ingreso = tabla_ingreso_intervalos.iloc[-1]
clase_moda_ingreso = tabla_ingreso_intervalos.loc[
    tabla_ingreso_intervalos["fi"].idxmax()
]

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Qué porcentaje tiene un ingreso mensual de 8 000 USD o más? "
    f"{porcentaje_ingreso_8000_mas:.1f}%."
)
print(
    "2. ¿Qué porcentaje cae en el último intervalo abierto? "
    f"{ultima_clase_ingreso['fi'] / n_ingreso * 100:.1f}% "
    f"({int(ultima_clase_ingreso['fi'])} personas)."
)
print(
    "Intervalo con mayor frecuencia:",
    clase_moda_ingreso["Intervalo"]
)
```

### 5.9 Tabla cruzada: tipo de usuario × propósito de uso

Se comparan estudiantes y profesionales según el propósito de uso. Se incluyen márgenes para conocer los totales por fila, por columna y el total general. Los faltantes de cualquiera de las dos variables no participan en el cruce.

**Interpretación:** esta tabla permite ver qué propósitos se relacionan con cada tipo de usuario y si la combinación más común representa una parte importante de todas las respuestas válidas.

```python
tabla_usuario_proposito = pd.crosstab(
    df_analisis["User_Type"],
    df_analisis["AI_Purpose"],
    margins=True,
    margins_name="Total",
    dropna=False
)

display(tabla_usuario_proposito)

n_cruce_usuario = df_analisis[
    ["User_Type", "AI_Purpose"]
].dropna().shape[0]
suma_cruce_usuario = int(
    tabla_usuario_proposito.loc[
        tabla_usuario_proposito.index != "Total",
        tabla_usuario_proposito.columns != "Total"
    ].to_numpy().sum()
)
print(
    f"Verificación del cruce: Σfi = {suma_cruce_usuario}; "
    f"n válido = {n_cruce_usuario}"
)
assert suma_cruce_usuario == n_cruce_usuario

cruce_sin_margenes = tabla_usuario_proposito.iloc[:-1, :-1]
combinacion_mas_frecuente = cruce_sin_margenes.stack().idxmax()
frecuencia_combinacion = cruce_sin_margenes.stack().max()
porcentaje_estudiantes_estudian = (
    tabla_usuario_proposito.loc["Student", "Studying"]
    / tabla_usuario_proposito.loc["Student", "Total"] * 100
)

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Cuál es la combinación más frecuente y sobre qué denominador? "
    f"{combinacion_mas_frecuente}: {int(frecuencia_combinacion)} "
    f"de {n_cruce_usuario} respuestas válidas."
)
print(
    "2. Entre estudiantes con propósito informado, ¿qué porcentaje "
    "indicó Studying? "
    f"{porcentaje_estudiantes_estudian:.1f}%."
)
```

### 5.10 Tabla cruzada con proporciones por fila: herramienta × disposición a recomendar

Se compara cada herramienta con la respuesta No, Maybe o Yes. Se usa `normalize="index"` para que cada fila represente proporciones dentro de los usuarios de esa herramienta. Se incluyen márgenes: la fila Total presenta la distribución general.

**Interpretación:** permite comparar la disposición a recomendar sin confundir el porcentaje con el número de usuarios. Una herramienta con pocos usuarios puede tener una proporción Yes alta, pero esa proporción debe leerse junto con el tamaño de su grupo.

```python
tabla_herramienta_recomendar = pd.crosstab(
    df_analisis["AI_Tool"],
    df_analisis["Would_Recommend"],
    margins=True,
    margins_name="Total",
    normalize="index",
    dropna=False
)

display((tabla_herramienta_recomendar * 100).round(1))

n_cruce_herramienta = df_analisis[
    ["AI_Tool", "Would_Recommend"]
].dropna().shape[0]
n_cruce_herramienta_check = int(
    df_analisis.dropna(
        subset=["AI_Tool", "Would_Recommend"]
    ).shape[0]
)
assert n_cruce_herramienta == n_cruce_herramienta_check
print(
    "El cruce usa", n_cruce_herramienta,
    "respuestas con herramienta y recomendación informadas."
)

proporcion_yes = tabla_herramienta_recomendar.drop(
    index="Total", errors="ignore"
)["Yes"].dropna()
herramienta_mayor_yes = proporcion_yes.idxmax()
n_herramienta_mayor_yes = int(
    df_analisis.loc[
        df_analisis["AI_Tool"] == herramienta_mayor_yes,
        "Would_Recommend"
    ].notna().sum()
)
porcentaje_yes_general = (
    tabla_herramienta_recomendar.loc["Total", "Yes"] * 100
)

print("\nPreguntas que se responden con la tabla:")
print(
    "1. ¿Qué herramienta tiene la mayor proporción de respuestas Yes? "
    f"{herramienta_mayor_yes} "
    f"({proporcion_yes.loc[herramienta_mayor_yes] * 100:.1f}%, "
    f"con {n_herramienta_mayor_yes} respuestas válidas en su fila)."
)
print(
    "2. ¿Qué porcentaje del total de respuestas válidas indicó Yes? "
    f"{porcentaje_yes_general:.1f}%."
)
```

### 6.2 Herramienta de IA: barras y Pareto

**Gráfico 3 — Barras.** Es apropiado para comparar una variable nominal con varias categorías. Las barras se ordenan de mayor a menor para facilitar la lectura y se mantienen las herramientas sin usuarios con frecuencia cero.

**Prompt de apoyo:** «Construye un gráfico de barras con el número de respuestas por AI_Tool, orden descendente, sin eliminar herramientas con cero registros y con eje vertical desde cero».

**Interpretación:** se aprecia qué herramientas destacan y cuánta distancia existe entre las más y menos usadas.

**Gráfico 4 — Pareto.** Combina barras ordenadas y porcentaje acumulado. Es útil para ver si unas pocas herramientas reúnen gran parte de las respuestas.

**Prompt de apoyo:** «Crea un diagrama de Pareto de AI_Tool: barras de frecuencia en orden descendente y línea de porcentaje acumulado con escala de 0 a 100».

**Interpretación:** la línea muestra qué proporción acumulada se alcanza al sumar las herramientas desde la más frecuente. Esto no se observa de forma tan directa en una tabla de frecuencias.

```python
frecuencia_herramienta = (
    df_analisis["AI_Tool"]
    .value_counts(sort=False)
    .reindex(df_analisis["AI_Tool"].cat.categories, fill_value=0)
)
frecuencia_herramienta_ordenada = frecuencia_herramienta.sort_values(
    ascending=False, kind="stable"
)

fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(
    frecuencia_herramienta_ordenada.index.astype(str),
    frecuencia_herramienta_ordenada.values
)
ax.set_title("Respuestas por herramienta de IA")
ax.set_xlabel("Herramienta")
ax.set_ylabel("Número de respuestas")
ax.set_ylim(bottom=0)
ax.tick_params(axis="x", rotation=35)
plt.tight_layout()
plt.show()

acumulada_pareto = (
    frecuencia_herramienta_ordenada.cumsum()
    / frecuencia_herramienta_ordenada.sum() * 100
)
fig, ax1 = plt.subplots(figsize=(10, 4))
posiciones = np.arange(len(frecuencia_herramienta_ordenada))
ax1.bar(posiciones, frecuencia_herramienta_ordenada.values)
ax1.set_xticks(posiciones)
ax1.set_xticklabels(
    frecuencia_herramienta_ordenada.index.astype(str),
    rotation=35,
    ha="right"
)
ax1.set_ylabel("Número de respuestas")
ax1.set_ylim(bottom=0)
ax1.set_xlabel("Herramienta")
ax1.set_title("Diagrama de Pareto de herramientas de IA")
ax2 = ax1.twinx()
ax2.plot(posiciones, acumulada_pareto.values, marker="o")
ax2.set_ylabel("Porcentaje acumulado")
ax2.set_ylim(0, 100)
ax2.set_yticks(np.arange(0, 101, 20))
plt.tight_layout()
plt.show()
```

### 6.3 Disposición a recomendar: barras ordenadas y porcentaje acumulado

**Gráfico 5 — Barras ordinales.** La variable es ordinal y por eso las barras respetan el orden No, Maybe y Yes, en vez de ordenarse por tamaño.

**Interpretación:** permite ver de inmediato si predominan respuestas favorables, intermedias o desfavorables.

**Gráfico 6 — Línea acumulada.** Se usa el orden de la escala para mostrar el porcentaje acumulado de No a Yes.

**Prompt de apoyo:** «Representa el porcentaje acumulado de la escala ordinal Would_Recommend como una línea escalonada; respeta el orden No, Maybe, Yes y usa eje de 0 a 100%».

**Interpretación:** muestra qué proporción se alcanza hasta cada nivel de la escala, algo que una tabla deja en cifras separadas.

```python
frecuencia_recomendacion = (
    df_analisis["Would_Recommend"]
    .value_counts(sort=False)
    .reindex(
        df_analisis["Would_Recommend"].cat.categories,
        fill_value=0
    )
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(
    frecuencia_recomendacion.index.astype(str),
    frecuencia_recomendacion.values
)
ax.set_title("Disposición a recomendar")
ax.set_xlabel("Respuesta")
ax.set_ylabel("Número de respuestas")
ax.set_ylim(bottom=0)
plt.tight_layout()
plt.show()

porcentaje_acumulado_recomendacion = (
    frecuencia_recomendacion.cumsum()
    / frecuencia_recomendacion.sum() * 100
)
fig, ax = plt.subplots(figsize=(7, 4))
ax.step(
    frecuencia_recomendacion.index.astype(str),
    porcentaje_acumulado_recomendacion.values,
    where="mid",
    marker="o"
)
ax.set_title("Porcentaje acumulado de disposición a recomendar")
ax.set_xlabel("Escala ordinal")
ax.set_ylabel("Porcentaje acumulado (%)")
ax.set_ylim(0, 100)
ax.set_yticks(np.arange(0, 101, 20))
plt.tight_layout()
plt.show()
```

### 6.4 Tipo de usuario: barras y barras apiladas al 100%

**Gráfico 8 — Barras apiladas al 100%.** Compara los propósitos dentro de cada tipo de usuario. La altura de cada segmento es un porcentaje de su fila.

## Producto 7. Medidas de tendencia central

Las medidas se calculan con los datos válidos de cada variable. La media utiliza todos los valores; la mediana se ubica en el centro de los datos ordenados; la moda identifica el valor o categoría que más se repite. Siempre se informa (n), porque las variables pueden tener cantidades diferentes de respuestas válidas.

### 7.1 Media, mediana y moda de variables cuantitativas

Se resumen siete variables: edad, horas de uso diario, horas ahorradas por semana, productividad, satisfacción, ingreso mensual y horas de trabajo o estudio por día. La moda puede tener más de un valor si hay empate.

```python
variables_cuantitativas = [
    "Age",
    "AI_Usage_Hours_Per_Day",
    "Time_Saved_Hours_Per_Week",
    "Productivity_Score",
    "Satisfaction_Score",
    "Monthly_Income",
    "Work_or_Study_Hours_Per_Day"
]

def texto_moda(serie):
    modas = serie.dropna().mode()
    if modas.empty:
        return "Sin datos"
    return ", ".join(str(valor) for valor in modas.tolist())

filas_tendencia = []
for columna in variables_cuantitativas:
    serie = pd.to_numeric(df_analisis[columna], errors="coerce").dropna()
    filas_tendencia.append({
        "Variable": columna,
        "n": len(serie),
        "Media": serie.mean(),
        "Mediana": serie.median(),
        "Moda": texto_moda(serie)
    })

tabla_tendencia = pd.DataFrame(filas_tendencia)
display(tabla_tendencia.round(3))

print(
    "Interpretación: cuando media y mediana son parecidas, "
    "el centro de los datos está en una zona cercana. "
    "Cuando se separan, puede existir asimetría o influencia de valores extremos."
)
```

### 7.2 Verificación de la suma de desviaciones respecto a la media

En una distribución numérica, la suma algebraica de las desviaciones respecto a su propia media debe ser cero. Se verifica con las horas de uso diario. Debido a la precisión de los decimales, se usa una tolerancia pequeña.

```python
serie_verificacion = df_analisis[
    "AI_Usage_Hours_Per_Day"
].dropna()

media_verificacion = serie_verificacion.mean()
suma_desviaciones = (
    serie_verificacion - media_verificacion
).sum()

print("n:", len(serie_verificacion))
print("Media:", media_verificacion)
print("Suma de desviaciones:", suma_desviaciones)
print(
    "¿La suma es cero dentro de la tolerancia?",
    np.isclose(suma_desviaciones, 0, atol=1e-8)
)
assert np.isclose(suma_desviaciones, 0, atol=1e-8)
```

### 7.3 Moda de variables nominales y mediana de variables ordinales

En las variables nominales solo se busca la moda, porque sus categorías no tienen orden. En las ordinales se calcula la mediana usando el orden declarado en el Producto 3. Cuando el número de respuestas es par, se muestran las dos categorías centrales; no se promedian sus nombres.

La media de una variable ordinal no tiene una interpretación clara porque la distancia entre categorías no está medida. Por ejemplo, no se puede asegurar que la diferencia entre No y Maybe sea igual a la diferencia entre Maybe y Yes.

```python
variables_nominales = [
    "Gender", "AI_Tool", "User_Type", "AI_Purpose"
]

filas_moda_nominal = []
for columna in variables_nominales:
    serie = df_analisis[columna].dropna()
    modas = serie.mode()
    filas_moda_nominal.append({
        "Variable": columna,
        "n": len(serie),
        "Moda": ", ".join(str(x) for x in modas.tolist())
    })

display(pd.DataFrame(filas_moda_nominal))


def mediana_ordinal(serie):
    categorias = list(serie.cat.categories)
    codigos = serie.cat.codes
    codigos = np.sort(codigos[codigos >= 0].to_numpy())
    n = len(codigos)

    if n == 0:
        return {"n": 0, "Mediana": "Sin datos", "Códigos centrales": ""}

    if n % 2 == 1:
        codigo = int(codigos[n // 2])
        texto = str(categorias[codigo])
        centrales = str(codigo)
    else:
        codigo_inferior = int(codigos[n // 2 - 1])
        codigo_superior = int(codigos[n // 2])
        if codigo_inferior == codigo_superior:
            texto = str(categorias[codigo_inferior])
        else:
            texto = (
                f"{categorias[codigo_inferior]} / "
                f"{categorias[codigo_superior]}"
            )
        centrales = f"{codigo_inferior} y {codigo_superior}"

    return {
        "n": n,
        "Mediana": texto,
        "Códigos centrales": centrales
    }


variables_ordinales = [
    "Education_Level", "Accuracy_Rating", "Would_Recommend"
]

filas_mediana_ordinal = []
for columna in variables_ordinales:
    resultado = mediana_ordinal(df_analisis[columna])
    filas_mediana_ordinal.append({
        "Variable": columna,
        **resultado
    })

display(pd.DataFrame(filas_mediana_ordinal))
```

### 7.4 Efecto de la limpieza: edad e ingreso mensual

Se comparan los mismos registros después de retirar duplicados. La columna «antes» convierte el texto a número, pero conserva los valores fuera de rango; la columna «después» utiliza los datos limpiados del Producto 2. Los valores imposibles se convierten en faltantes, no se reemplazan por números inventados.

**Interpretación:** si la media cambia más que la mediana, el dato extremo estaba influyendo especialmente en la media. La mediana suele ser menos sensible a unos pocos valores alejados.

```python
def convertir_sin_tratar_rangos(serie):
    texto = (
        serie.astype("string")
        .str.replace("yrs", "", regex=False)
        .str.replace("hrs", "", regex=False)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip()
    )
    return pd.to_numeric(texto, errors="coerce")


base_antes_limpieza = df_encuesta.drop_duplicates().copy()
filas_efecto_limpieza = []

for columna in ["Age", "Monthly_Income"]:
    antes = convertir_sin_tratar_rangos(
        base_antes_limpieza[columna]
    ).dropna()
    despues = pd.to_numeric(
        df_normalizada[columna], errors="coerce"
    ).dropna()

    filas_efecto_limpieza.append({
        "Variable": columna,
        "n antes": len(antes),
        "Media antes": antes.mean(),
        "Mediana antes": antes.median(),
        "n después": len(despues),
        "Media después": despues.mean(),
        "Mediana después": despues.median()
    })

comparacion_limpieza = pd.DataFrame(filas_efecto_limpieza)
display(comparacion_limpieza.round(2))

print(
    "La comparación muestra el efecto de convertir los valores fuera "
    "de rango en faltantes. La mediana permite contrastar cuánto se "
    "movió el centro sin depender tanto del valor extremo."
)
```

### 7.7 Trampa del promedio de promedios: productividad por país

Se calcula la media de productividad de cada país. Después se compara el promedio simple de esas medias con la media general ponderada por el número de respuestas válidas de cada país.

**Interpretación:** el promedio simple da el mismo peso a un país con pocas respuestas y a uno con muchas. La media ponderada considera el tamaño de cada grupo y coincide con la media general de los registros válidos.

```python
productividad_por_pais = (
    df_analisis.dropna(subset=["Country", "Productivity_Score"])
    .groupby("Country", observed=True)
    .agg(
        n=("Productivity_Score", "count"),
        Media_productividad=("Productivity_Score", "mean")
    )
    .reset_index()
)

promedio_simple_medias = productividad_por_pais[
    "Media_productividad"
].mean()

promedio_ponderado = (
    productividad_por_pais["Media_productividad"]
    * productividad_por_pais["n"]
).sum() / productividad_por_pais["n"].sum()

media_general = df_analisis["Productivity_Score"].mean()

display(productividad_por_pais.sort_values("n", ascending=False).round(2))
print("Promedio simple de las medias por país:", round(promedio_simple_medias, 3))
print("Promedio ponderado por n:", round(promedio_ponderado, 3))
print("Media general de los registros:", round(media_general, 3))
print(
    "Diferencia entre promedio simple y media general:",
    round(promedio_simple_medias - media_general, 3)
)
assert np.isclose(promedio_ponderado, media_general, atol=1e-10)
```

## Producto 8. Medidas de dispersión

Las medidas de dispersión muestran qué tanto se separan las respuestas entre sí. Se informan los tamaños de muestra porque una variable puede tener más datos válidos que otra. Se diferencia la varianza y desviación estándar muestral ((ddof=1)) de la poblacional ((ddof=0)).

### 8.1 Rango, varianza, desviación estándar y coeficiente de variación

Se calculan estas medidas para seis variables cuantitativas. El rango es máximo menos mínimo. La varianza y desviación estándar muestrales usan (n-1); las poblacionales usan (n). El CV se calcula como desviación estándar muestral dividida por la media y multiplicada por 100.

La versión muestral se usa cuando las respuestas se consideran una muestra de un grupo más amplio. La versión poblacional describe únicamente el conjunto completo que se está analizando. En este trabajo se muestran ambas para dejar clara la diferencia.

```python
variables_dispersion = [
    "Age",
    "Monthly_Income",
    "AI_Usage_Hours_Per_Day",
    "Time_Saved_Hours_Per_Week",
    "Productivity_Score",
    "Satisfaction_Score"
]

filas_dispersion = []
for columna in variables_dispersion:
    serie = pd.to_numeric(df_analisis[columna], errors="coerce").dropna()
    media = serie.mean()
    desviacion_muestral = serie.std(ddof=1)
    filas_dispersion.append({
        "Variable": columna,
        "n": len(serie),
        "Mínimo": serie.min(),
        "Máximo": serie.max(),
        "Rango": serie.max() - serie.min(),
        "Varianza muestral (ddof=1)": serie.var(ddof=1),
        "Varianza poblacional (ddof=0)": serie.var(ddof=0),
        "Desv. estándar muestral": desviacion_muestral,
        "Desv. estándar poblacional": serie.std(ddof=0),
        "CV muestral (%)": (
            desviacion_muestral / media * 100
            if media != 0 else np.nan
        )
    })

tabla_dispersion = pd.DataFrame(filas_dispersion)
display(tabla_dispersion.round(3))

print(
    "La varianza y desviación muestral son un poco mayores porque "
    "dividen por n-1. La versión poblacional divide por n."
)
```

### 8.2 Comparación del CV entre variables y entre grupos

El coeficiente de variación permite comparar dispersión relativa aunque las variables tengan unidades distintas, siempre que tengan un cero real y una media positiva. Se compara ingreso mensual con horas de uso diario y, luego, horas ahorradas entre estudiantes y profesionales y entre las cuatro herramientas más usadas.

```python
cv_entre_variables = tabla_dispersion.set_index("Variable")[
    "CV muestral (%)"
].reindex([
    "Monthly_Income",
    "AI_Usage_Hours_Per_Day"
])
display(cv_entre_variables.to_frame("CV muestral (%)").round(2))

def resumen_cv_por_grupo(df, columna_grupo, variable):
    salida = (
        df.groupby(columna_grupo, observed=True)[variable]
        .agg(n="count", media="mean", desviacion="std")
        .reset_index()
    )
    salida["CV (%)"] = (
        salida["desviacion"] / salida["media"] * 100
    )
    return salida

cv_ahorro_usuario = resumen_cv_por_grupo(
    df_analisis, "User_Type", "Time_Saved_Hours_Per_Week"
)
print("Horas ahorradas por tipo de usuario:")
display(cv_ahorro_usuario.round(2))

cv_ahorro_herramienta = resumen_cv_por_grupo(
    df_analisis[
        df_analisis["AI_Tool"].isin(herramientas_mas_usadas)
    ],
    "AI_Tool",
    "Time_Saved_Hours_Per_Week"
).sort_values("n", ascending=False)

print("Horas ahorradas por herramienta (cuatro más usadas):")
display(cv_ahorro_herramienta.round(2))

print(
    "El grupo con CV más alto tiene mayor dispersión relativa. "
    "Se debe leer junto al n: un CV calculado con pocas respuestas "
    "puede ser menos estable."
)
```

### 8.3 Cuartiles por el método de las mitades y comparación con pandas

Se calculan Q1 y Q3 ordenando los datos y dividiéndolos en dos mitades. Si (n) es impar, el dato central se deja fuera de las dos mitades; si (n) es par, cada mitad contiene la misma cantidad de valores. Luego se compara con `quantile()` de pandas, que por defecto interpola linealmente.

Para cada variable se calcula el RIC, las cercas (Q1 - 1.5\cdot RIC) y (Q3 + 1.5\cdot RIC), y la lista de observaciones fuera de esas cercas. Se revisan edad, ingreso mensual y horas ahorradas.

```python
def mediana_lista_ordenada(valores):
    valores = np.sort(np.asarray(valores, dtype=float))
    n = len(valores)
    if n == 0:
        return np.nan
    if n % 2 == 1:
        return float(valores[n // 2])
    return float((valores[n // 2 - 1] + valores[n // 2]) / 2)


def cuartiles_metodo_mitades(serie):
    valores = np.sort(
        pd.to_numeric(serie, errors="coerce").dropna().to_numpy()
    )
    n = len(valores)

    if n == 0:
        return np.nan, np.nan, np.nan

    mitad = n // 2
    if n % 2 == 1:
        mitad_inferior = valores[:mitad]
        mitad_superior = valores[mitad + 1:]
    else:
        mitad_inferior = valores[:mitad]
        mitad_superior = valores[mitad:]

    q1 = mediana_lista_ordenada(mitad_inferior)
    q2 = mediana_lista_ordenada(valores)
    q3 = mediana_lista_ordenada(mitad_superior)
    return q1, q2, q3


variables_cuartiles = [
    "Age",
    "Monthly_Income",
    "Time_Saved_Hours_Per_Week"
]

filas_cuartiles = []
valores_atipicos = {}

for columna in variables_cuartiles:
    serie = pd.to_numeric(df_analisis[columna], errors="coerce").dropna()
    q1, mediana, q3 = cuartiles_metodo_mitades(serie)
    ric = q3 - q1
    limite_inferior = q1 - 1.5 * ric
    limite_superior = q3 + 1.5 * ric

    q1_pandas = serie.quantile(0.25)
    q3_pandas = serie.quantile(0.75)

    mascara_atipicos = (
        (serie < limite_inferior) | (serie > limite_superior)
    )
    ids_atipicos = df_analisis.loc[
        mascara_atipicos.reindex(df_analisis.index, fill_value=False),
        "User_ID"
    ]
    datos_atipicos = pd.DataFrame({
        "User_ID": ids_atipicos,
        columna: df_analisis.loc[ids_atipicos.index, columna]
    })
    valores_atipicos[columna] = datos_atipicos

    filas_cuartiles.append({
        "Variable": columna,
        "n": len(serie),
        "Q1 mitades": q1,
        "Mediana": mediana,
        "Q3 mitades": q3,
        "RIC mitades": ric,
        "Límite inferior": limite_inferior,
        "Límite superior": limite_superior,
        "Q1 pandas": q1_pandas,
        "Q3 pandas": q3_pandas,
        "Q1 diferencia": q1 - q1_pandas,
        "Q3 diferencia": q3 - q3_pandas,
        "Cantidad de atípicos": int(mascara_atipicos.sum())
    })

tabla_cuartiles = pd.DataFrame(filas_cuartiles)
display(tabla_cuartiles.round(3))

for columna, tabla_atipicos in valores_atipicos.items():
    print(f"Valores atípicos de {columna}:")
    if tabla_atipicos.empty:
        print("No se encontraron valores fuera de las cercas.")
    else:
        display(tabla_atipicos.head(25))
        if len(tabla_atipicos) > 25:
            print("Se muestran los primeros 25 de", len(tabla_atipicos))
```

### 8.6 Robustez: efecto de un valor con un cero de más

Se toma una respuesta de horas ahorradas cercana a la mediana y se simula que se digitó con un cero adicional, multiplicándola por diez. Se compara la desviación estándar y el RIC antes y después. La simulación se hace sobre una copia y no modifica los datos de la encuesta.

**Interpretación:** la desviación estándar suele reaccionar mucho ante un valor extremo porque usa las distancias respecto a la media. El RIC depende de la zona central y suele cambiar menos cuando solo se altera una observación.

```python
serie_original = (
    df_analisis["Time_Saved_Hours_Per_Week"]
    .dropna()
    .copy()
)

valor_central = serie_original.median()
indice_modificado = serie_original[
    serie_original == valor_central
].index[0]

serie_simulada = serie_original.copy()
valor_original = serie_simulada.loc[indice_modificado]
serie_simulada.loc[indice_modificado] = valor_original * 10


def resumen_robustez(serie):
    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)
    return {
        "n": len(serie),
        "Media": serie.mean(),
        "Desviación estándar": serie.std(ddof=1),
        "Q1": q1,
        "Q3": q3,
        "RIC": q3 - q1
    }


comparacion_robustez = pd.DataFrame([
    {
        "Escenario": "Datos originales",
        **resumen_robustez(serie_original)
    },
    {
        "Escenario": "Un valor con un cero de más",
        **resumen_robustez(serie_simulada)
    }
])

print(
    "User_ID del dato modificado:",
    df_analisis.loc[indice_modificado, "User_ID"]
)
print("Valor original:", valor_original)
print("Valor simulado:", serie_simulada.loc[indice_modificado])
display(comparacion_robustez.round(3))

cambio_desviacion = (
    comparacion_robustez.loc[1, "Desviación estándar"]
    - comparacion_robustez.loc[0, "Desviación estándar"]
)
cambio_ric = (
    comparacion_robustez.loc[1, "RIC"]
    - comparacion_robustez.loc[0, "RIC"]
)
print("Cambio en desviación estándar:", round(cambio_desviacion, 3))
print("Cambio en RIC:", round(cambio_ric, 3))
print(
    "La comparación muestra que una sola digitación extrema puede "
    "alterar mucho la desviación estándar, mientras el RIC suele "
    "mantenerse más estable."
)
```
