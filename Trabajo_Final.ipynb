### 2.2 Crear la copia de trabajo y retirar duplicados

### 2.3 Estandarizar las categorías

Se quitan espacios sobrantes y se reemplazan las variantes que ya se detectaron en el Producto 1. Los nombres finales se toman de los archivos de referencia del repositorio: el Excel de herramientas y el JSON de países.

No se rellenan las celdas vacías. Solo se cambian los textos para los que existe una equivalencia clara.

```python
columnas_texto = ["Gender", "Country", "Education_Level", "AI_Tool"]

for columna in columnas_texto:
    df_normalizada[columna] = df_normalizada[columna].str.strip()

mapa_genero = {
    "M": "Male",
    "male": "Male",
    "MALE": "Male",
    "F": "Female",
    "female": "Female",
    "FEMALE": "Female",
    "other": "Other"
}

mapa_pais = {
    "australia": "Australia",
    "brazil": "Brazil",
    "canada": "Canada",
    "france": "France",
    "india": "India",
    "philippines": "Philippines",
    "U.K.": "United Kingdom",
    "United States of America": "United States"
}

mapa_educacion = {
    "BA/BSc": "Bachelor's",
    "bachelor's degree": "Bachelor's",
    "Bachelors": "Bachelor's",
    "high school": "High School",
    "HS": "High School",
    "MA/MSc": "Master's",
    "master's degree": "Master's",
    "Masters": "Master's",
    "Ph.D.": "PhD",
    "undergrad": "Undergraduate"
}

mapa_herramientas = {
    "Chat GPT": "ChatGPT",
    "chatgpt": "ChatGPT",
    "Chatgpt": "ChatGPT",
    "claude": "Claude",
    "gemini": "Google Gemini",
    "Copilot": "Microsoft Copilot",
    "copilot": "Microsoft Copilot"
}

df_normalizada["Gender"] = df_normalizada["Gender"].replace(mapa_genero)
df_normalizada["Country"] = df_normalizada["Country"].replace(mapa_pais)
df_normalizada["Education_Level"] = df_normalizada["Education_Level"].replace(mapa_educacion)
df_normalizada["AI_Tool"] = df_normalizada["AI_Tool"].replace(mapa_herramientas)

print("Valores de Gender:", sorted(df_normalizada["Gender"].dropna().unique()))
print("Valores de Country:", sorted(df_normalizada["Country"].dropna().unique()))
print("Valores de Education_Level:", sorted(df_normalizada["Education_Level"].dropna().unique()))
print("Valores de AI_Tool:", sorted(df_normalizada["AI_Tool"].dropna().unique()))
```

### 2.4 Convertir números y tratar los valores fuera de rango

Algunos números llegaron como texto porque tenían unidades o símbolos. La función \`a_numero()\` retira esos caracteres y usa \`pd.to_numeric()\` para convertir la columna. Después se aplican los rangos establecidos en el diagnóstico.

Cuando un número queda fuera del rango, se cambia solamente esa celda a \`NaN\`; no se elimina toda la fila.

```python
columnas_numericas_texto = [
    "Age", "Monthly_Income", "AI_Usage_Hours_Per_Day",
    "Monthly_AI_Cost", "Time_Saved_Hours_Per_Week",
    "Work_or_Study_Hours_Per_Day"
]

def a_numero(serie):
    limpio = (serie.str.replace("yrs", "", regex=False)
                    .str.replace("hrs", "", regex=False)
                    .str.replace("$", "", regex=False)
                    .str.replace(",", "", regex=False)
                    .str.strip())
    return pd.to_numeric(limpio, errors="coerce")

for columna in columnas_numericas_texto:
    df_normalizada[columna] = a_numero(df_normalizada[columna])

rangos_validos = {
    "Age": (16, 80),
    "Monthly_Income": (0, 20000),
    "AI_Usage_Hours_Per_Day": (0, 16),
    "Work_or_Study_Hours_Per_Day": (0, 16),
    "Time_Saved_Hours_Per_Week": (0, 168),
    "Productivity_Score": (1, 10),
    "Accuracy_Rating": (1, 5),
    "Satisfaction_Score": (1, 10)
}

for columna, limites in rangos_validos.items():
    fuera_de_rango = (df_normalizada[columna].notna() &
                      ~df_normalizada[columna].between(limites[0], limites[1]))
    print(columna, "- valores fuera de rango:", fuera_de_rango.sum())
    df_normalizada.loc[fuera_de_rango, columna] = np.nan

costos_negativos = (df_normalizada["Monthly_AI_Cost"].notna() &
                    (df_normalizada["Monthly_AI_Cost"] < 0))
print("Monthly_AI_Cost - valores negativos:", costos_negativos.sum())
df_normalizada.loc[costos_negativos, "Monthly_AI_Cost"] = np.nan
```

### 2.5 Crear la columna \`Num_Tareas\`

En \`Tasks_Performed\` las tareas están separadas por comas. Se divide el texto por ese separador y se cuenta cuántos fragmentos hay.

Cuando \`Tasks_Performed\` está vacío, \`Num_Tareas\` también queda como faltante. No se asigna cero porque no sabemos cuántas tareas realiza esa persona.

```python
df_normalizada["Num_Tareas"] = (
    df_normalizada["Tasks_Performed"]
    .str.split(",")
    .str.len()
    .astype("Int64")
)

print("Cantidad de tareas registradas:")
print(df_normalizada["Num_Tareas"].value_counts(dropna=False).sort_index())
```

### 2.6 Crear la columna \`Area_Profesional\`

La profesión es texto libre y aparece en varios idiomas. Para resumirla, se crean familias a partir de palabras clave con \`str.contains()\`.

**Orden de las reglas:** primero se identifican los estudiantes; luego las profesiones de desarrollo, datos, atención al cliente, investigación, contenido y administración. El orden evita que un estudiante sea clasificado como profesional por una palabra que aparezca en su descripción.

Las profesiones que no coincidan con ninguna regla quedan en **Otra**. Cuando \`Profession\` está vacío, \`Area_Profesional\` también queda como faltante: no se inventa una profesión.

```python
profesion = df_normalizada["Profession"].str.strip()

es_estudiante = profesion.str.contains(
    r"\bstudent\b", case=False, na=False
)

es_estudiante_ingenieria = profesion.str.contains(
    r"computer science|software|engineering|engineer|information technology|\bIT\b|machine learning|artificial intelligence|data science|electronics|computing|programming",
    case=False, na=False
)

df_normalizada["Area_Profesional"] = "Otra"

# Los estudiantes se clasifican antes que las profesiones.
df_normalizada.loc[
    es_estudiante & es_estudiante_ingenieria,
    "Area_Profesional"
] = "Estudiante de ingeniería y computación"

df_normalizada.loc[
    es_estudiante & ~es_estudiante_ingenieria,
    "Area_Profesional"
] = "Estudiante de otras áreas"

no_estudiante = profesion.notna() & ~es_estudiante

# 1. Desarrollo de software
mascara = no_estudiante & profesion.str.contains(
    r"software\s+(developer|engineer|entwickler|entwicklerin)|softwareentwickler(in)?|developer|desenvolvedor|desenvolvedora|développeur|logiciel|full\s*stack|tech lead|technical lead|machine learning engineer|systems developer",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Desarrollo de software"

# 2. Análisis de datos
mascara = no_estudiante & profesion.str.contains(
    r"data analyst|data scientist|data science|data analytics|business intelligence|business analytics|analyste de données|analista de dados|datenanalyst|analytics",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Análisis de datos"

# 3. Atención al cliente
mascara = no_estudiante & profesion.str.contains(
    r"customer|support|service|success|kundenservice|kundenbetreuer|atendimento",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Atención al cliente"

# 4. Investigación y academia
mascara = no_estudiante & profesion.str.contains(
    r"research|professor|academic|scientist|researcher|pesquisador|phd",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Investigación y academia"

# 5. Contenido y marketing
mascara = no_estudiante & profesion.str.contains(
    r"content|contenu|writer|writing|marketing|social media|digital media|multimedia|film production|gestor de conteúdo|copywriting",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Contenido y marketing"

# 6. Administración y negocios
mascara = no_estudiante & profesion.str.contains(
    r"administrat|business analyst|business administration|project manager|projektmanager|data entry|assistente administrativo",
    case=False, na=False
)
df_normalizada.loc[mascara, "Area_Profesional"] = "Administración y negocios"

# Los cargos sin coincidencia quedan como "Otra".
# Los 14 valores vacíos de Profession se conservan como faltantes.
df_normalizada.loc[
    df_normalizada["Profession"].isna(),
    "Area_Profesional"
] = np.nan

frecuencia_areas = df_normalizada["Area_Profesional"].value_counts(
    dropna=False
).to_frame("Registros")
frecuencia_areas["% del total"] = (
    frecuencia_areas["Registros"] / len(df_normalizada) * 100
).round(1)

display(frecuencia_areas)

registros_otra = df_normalizada["Area_Profesional"].eq("Otra").sum()
porcentaje_otra = registros_otra / len(df_normalizada) * 100

print("Registros en Otra:", registros_otra)
print(f"Porcentaje de registros en Otra: {porcentaje_otra:.1f}%")
print("Profesiones sin información:", df_normalizada["Profession"].isna().sum())
```
